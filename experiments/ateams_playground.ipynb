```python
from ateams.complexes import Cubical
from ateams.models import InvadedCluster, Bernoulli
from ateams.statistics import critical
from ateams import Chain, Recorder
```

```python
field = 2
C = Cubical().fromCorners([2**8]*2)
IC = InvadedCluster(C, dimension=1, field=field)
print(C.vertexMap)
print(C.corners)
with Recorder().record("out.lz") as rec:
  for (spins, occupied, satisfied) in Chain(IC, steps=100).progress():
    rec.store((spins, occupied, satisfied))
```

```python
from ateams import Player
from ateams.statistics import occupancy

sum_occupied = [0]
sum_total = [0]

with Player().playback("out.lz", steps=1) as play:
  for (spins, occupied, satisfied) in play.progress():
    print(occupancy((spins, occupied, satisfied)))
    print(occupancy(((spins, occupied, satisfied)))/len(occupied))
    sum_occupied.append(occupancy((spins, occupied, satisfied)) + sum_occupied[-1])
    sum_total.append(len(occupied) + sum_total[-1])
  
print(sum_occupied)
print(sum_total)
```
